```python
from ultralytics import YOLO
import yaml
import os
```

## 1. Verify Dataset configuration
Point YOLO to the `data.yaml` from Roboflow.


```python
DATA_YAML = 'data.yaml'

if os.path.exists(DATA_YAML):
    with open(DATA_YAML, 'r') as f:
        data_config = yaml.safe_load(f)
    print(data_config)
else:

    # Only when notebook 01 doesnt run first
    print(f'Error: {DATA_YAML} not found.')
```

## 2. Train YOLOv8 Nano (Baseline)


```python
# Load YOLOv11 nano
model = YOLO("yolo11n.pt")

results = model.train(
    data=DATA_YAML,
    epochs=20,
    imgsz=640,
    project="yolo_highway",
    name="yolo11_fast",
    cache=False,       # Dont cache images in RAM
    half=True,
    amp=True,
    mosaic=1.0,
    mixup=0.2,
    copy_paste=0.1,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4
)
```

```python
model = YOLO("best.pt")
```

```python
# Evaluate model
metrics = model.val(
    data="data.yaml",
    imgsz=640,
    split="val",
    plots=True
)
```
